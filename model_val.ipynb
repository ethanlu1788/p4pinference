```python
from ultralytics import YOLO

# Load PyTorch model
model = YOLO("models/best_openvino_model")

# Validate on your custom dataset
metrics = model.val(data="val/val.yaml")

print(metrics.box.map)    # mAP 0.5:0.95
print(metrics.box.map50)  # mAP@.50
print(metrics.box.map75)  # mAP@.75
print(metrics.box.maps)   # mAP for each class
```
